```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("Employee_Attrition_Analysis/Dataset/employee_attrition.csv")

print("Dataset loaded successfully!")
```

```python
print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Total Missing Values:", df.isnull().sum().sum())
```

```python
print("Duplicate Records:", df.duplicated().sum())
```

```python
df.describe()
```

```python
print("Employee Attrition Count:")
print(df['Attrition'].value_counts())
```

```python
total_employees = len(df)

employees_left = (df['Attrition'] == 'Yes').sum()

attrition_rate = (employees_left / total_employees) * 100

print("Total Employees:", total_employees)
print("Employees Left:", employees_left)
print("Attrition Rate:", round(attrition_rate, 2), "%")
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x='Attrition')

plt.title("Employee Attrition Count")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Department',
    hue='Attrition'
)

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
```

```python
department_attrition = pd.crosstab(
    df['Department'],
    df['Attrition']
)

department_attrition['Attrition Rate'] = (
    department_attrition['Yes'] /
    (department_attrition['Yes'] + department_attrition['No'])
) * 100

print(department_attrition)
```

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[18, 25, 35, 45, 55, 65],
    labels=['18-25', '26-35', '36-45', '46-55', '56-65']
)

print(df[['Age', 'Age_Group', 'Attrition']].head(10))
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='Age_Group',
    hue='Attrition'
)

plt.title("Employee Attrition by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Employees")

plt.show()
```

```python
age_attrition = pd.crosstab(
    df['Age_Group'],
    df['Attrition']
)

age_attrition['Attrition Rate'] = (
    age_attrition['Yes'] /
    (age_attrition['Yes'] + age_attrition['No'])
) * 100

print(age_attrition.round(2))
```

```python
print(df['MonthlyIncome'].describe())
```

```python
df['Salary_Group'] = pd.cut(
    df['MonthlyIncome'],
    bins=[0, 3000, 6000, 10000, 20000],
    labels=['Low Salary', 'Medium Salary', 'High Salary', 'Very High Salary']
)

print(df[['MonthlyIncome', 'Salary_Group', 'Attrition']].head(10))
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='Salary_Group',
    hue='Attrition'
)

plt.title("Employee Attrition by Salary Group")
plt.xlabel("Salary Group")
plt.ylabel("Number of Employees")

plt.show()
```

```python
salary_attrition = pd.crosstab(
    df['Salary_Group'],
    df['Attrition']
)

salary_attrition['Attrition Rate'] = (
    salary_attrition['Yes'] /
    (salary_attrition['Yes'] + salary_attrition['No'])
) * 100

print(salary_attrition.round(2))
```

```python
satisfaction_attrition = pd.crosstab(
    df['JobSatisfaction'],
    df['Attrition']
)

satisfaction_attrition['Attrition Rate'] = (
    satisfaction_attrition['Yes'] /
    (satisfaction_attrition['Yes'] +
     satisfaction_attrition['No'])
) * 100

print(satisfaction_attrition.round(2))
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Department-wise attrition
sns.countplot(
    data=df,
    x='Department',
    hue='Attrition',
    ax=axes[0, 0]
)
axes[0, 0].set_title("Attrition by Department")

# 2. Age-wise attrition
sns.countplot(
    data=df,
    x='Age_Group',
    hue='Attrition',
    ax=axes[0, 1]
)
axes[0, 1].set_title("Attrition by Age")

# 3. Salary-wise attrition
sns.countplot(
    data=df,
    x='Salary_Group',
    hue='Attrition',
    ax=axes[1, 0]
)
axes[1, 0].set_title("Attrition by Salary")

# 4. Overtime-wise attrition
sns.countplot(
    data=df,
    x='OverTime',
    hue='Attrition',
    ax=axes[1, 1]
)
axes[1, 1].set_title("Attrition by Overtime")

plt.tight_layout()
plt.show()
```
